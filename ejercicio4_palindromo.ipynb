# Ejercicio 4: Palíndromo
Identifica si una cadena introducida por teclado es un palíndromo (ejemplo: 'oso' es palíndromo).

```python
def es_palindromo(cadena):
    cadena_limpia = cadena.lower().replace(" ", "")
    return cadena_limpia == cadena_limpia[::-1]

cadena = input("Ingrese una palabra o frase: ")
if es_palindromo(cadena):
    print(f"'{cadena}' SÍ es un palíndromo.")
else:
    print(f"'{cadena}' NO es un palíndromo.")
```
