```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

img_color = cv2.imread('airplane.jpg')
img_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_color, cv2.COLOR_RGB2GRAY)

kernels = [
    ("desenfoque", 
     np.array([[1, 1, 1],
               [1, 1, 1],
               [1, 1, 1]], dtype=np.float32) / 9.0),

    ("filtro gauss 5x5", 
     np.array([[1,  4,  7,  4, 1],
               [4, 16, 26, 16, 4],
               [7, 26, 41, 26, 7],
               [4, 16, 26, 16, 4],
               [1,  4,  7,  4, 1]], dtype=np.float32) / 273.0),

    ("detección bordes", 
     np.array([[ 0,  1,  0],
               [ 1, -4,  1],
               [ 0,  1,  0]], dtype=np.float32)),

    ("sobel horizontal", 
     np.array([[-1, 0, 1],
               [-2, 0, 2],
               [-1, 0, 1]], dtype=np.float32)),

    ("realce bordes", 
     np.array([[ 0,  0, 0],
               [-1,  1, 0],
               [ 0,  0, 0]], dtype=np.float32)),

    ("enfoque", 
     np.array([[ 0, -1,  0],
               [-1,  5, -1],
               [ 0, -1,  0]], dtype=np.float32)),

    ("sharpen", 
     np.array([[ 1, -2,  1],
               [-2,  5, -2],
               [ 1, -2,  1]], dtype=np.float32)),

    ("filtro norte", 
     np.array([[ 1,  1,  1],
               [ 1, -2,  1],
               [-1, -1, -1]], dtype=np.float32)),

    ("filtro este", 
     np.array([[-1, 1, 1],
               [-1, -2, 1],
               [-1, 1, 1]], dtype=np.float32)),

    ("repujado", 
     np.array([[-2, -1, 0],
               [-1,  1, 1],
               [ 0,  1, 2]], dtype=np.float32)),
]


def mostrar_resultados(imagen, es_gris=False):
    """Aplica todos los kernels a una imagen y muestra los resultados."""
    fig, axes = plt.subplots(2, 6, figsize=(18, 6))
    axes = axes.flatten()
    
    cmap = 'gray' if es_gris else None
    axes[0].imshow(imagen, cmap=cmap)
    axes[0].set_title('Original')
    axes[0].axis('off')
    
    for i, (nombre, kernel) in enumerate(kernels, start=1):
        salida = cv2.filter2D(imagen, -1, kernel)
        axes[i].imshow(salida, cmap=cmap)
        axes[i].set_title(nombre)
        axes[i].axis('off')
    
    axes[-1].axis('off')
    
    plt.tight_layout()
    plt.show()


mostrar_resultados(img_gray, es_gris=True)
mostrar_resultados(img_color, es_gris=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_media_h(imagen, tkernel):
    kernel=np.ones(tkernel, np.float32)/np.prod(tkernel)
    imagen_c=cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img= cv2.imread('gurl.jpeg', 0)  

img_fin= convolucion_media_h(img, 15)
plt.figure(figsize=(10,5))
plt.subplot(121)

plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.imshow(img_fin,cmap='gray')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((tkernel, tkernel), np.float32) / (tkernel * tkernel)
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img = cv2.imread('gurl.jpeg')


img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


img_fin = convolucion_media_h(img_rgb, 15)


plt.figure(figsize=(10,5))

plt.subplot(121)
plt.title("Original")
plt.imshow(img_rgb)

plt.subplot(122)
plt.title("Filtrada")
plt.imshow(img_fin)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2


def crear_kernel_direccional(tamano, direccion):
    """Crea un kernel de desenfoque direccional."""
    kernel = np.zeros((tamano, tamano), dtype=np.float32)
    
    if direccion == 'horizontal':
        kernel[tamano // 2, :] = 1
    elif direccion == 'vertical':
        kernel[:, tamano // 2] = 1
    elif direccion == 'diagonal':
        np.fill_diagonal(kernel, 1)
    elif direccion == 'diagonal_inversa':
        np.fill_diagonal(np.fliplr(kernel), 1)
    else:
        raise ValueError(f"Dirección no válida: {direccion}")
    
    return kernel / kernel.sum()


img = cv2.imread('airplane.jpg')
if img is None:
    raise FileNotFoundError("No se pudo cargar 'airplane.jpg'")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

tam_kernel = 15
direcciones = ['horizontal', 'vertical', 'diagonal', 'diagonal_inversa']
titulos = ['Horizontal', 'Vertical', 'Diagonal', 'Diagonal Inversa']

resultados = [cv2.filter2D(img_rgb, -1, crear_kernel_direccional(tam_kernel, d)) 
              for d in direcciones]

fig, axes = plt.subplots(1, 5, figsize=(15, 4))

axes[0].imshow(img_rgb)
axes[0].set_title('Original')
axes[0].axis('off')

for ax, img_res, titulo in zip(axes[1:], resultados, titulos):
    ax.imshow(img_res)
    ax.set_title(titulo)
    ax.axis('off')

plt.tight_layout()
plt.show()
```
