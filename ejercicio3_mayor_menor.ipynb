# Ejercicio 3: Mayor y Menor
Permite ingresar diferentes números con un límite definido inicialmente (no menor a 3) y determina cuál es el mayor y el menor.

```python
try:
    limite = int(input("¿Cuántos números desea ingresar? (mínimo 3): "))
    if limite < 3:
        print("El límite debe ser al menos 3. Se usará 3 por defecto.")
        limite = 3

    numeros = []
    for i in range(1, limite + 1):
        num = float(input(f"Ingrese el número {i}: "))
        numeros.append(num)

    mayor = numeros[0]
    menor = numeros[0]
    for num in numeros[1:]:
        if num > mayor:
            mayor = num
        if num < menor:
            menor = num

    print(f"\nNúmeros ingresados: {numeros}")
    print(f"El número MAYOR es: {mayor}")
    print(f"El número MENOR es: {menor}")
except ValueError:
    print("Por favor, ingrese valores numéricos válidos.")
```
