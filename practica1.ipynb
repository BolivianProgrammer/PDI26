Mis imagenes

```python
import cv2 
import matplotlib.pyplot as plt
import numpy as np
img1 = cv2.imread('pose1.jpeg')
img_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
plt.imsave('pose1_rgb.jpeg', img_rgb)
fig, axs = plt.subplots(1,4, figsize=(20,5))
axs[0].imshow(img_rgb)
axs[0].set_title("Primera pose")

fig_img = axs[1].imshow(img_rgb[..., 0], cmap='gray')
fig.colorbar(fig_img, ax=axs[1], shrink=0.5)
axs[1].set_title('Red Channel')
plt.imsave('pose1_red_channel.jpeg', img_rgb[..., 0], cmap='gray')

fig_img = axs[2].imshow(img_rgb[..., 1], cmap='gray')
fig.colorbar(fig_img, ax=axs[2], shrink=0.5)
axs[2].set_title('Green Channel')
plt.imsave('pose1_green_channel.jpeg', img_rgb[..., 1], cmap='gray')

fig_img = axs[3].imshow(img_rgb[..., 2], cmap='gray')
fig.colorbar(fig_img, ax=axs[3], shrink=0.5)
axs[3].set_title('Blue Channel')
plt.imsave('pose1_blue_channel.jpeg', img_rgb[..., 2], cmap='gray')
```

Funcion para mostrar las imagenes guardadas

```python
import cv2
import matplotlib.pyplot as plt

def show_poses():
    poses = ['pose1.jpeg', 'pose2.jpeg', 'pose3.jpeg']
    titles = ['Primera pose', 'Segunda pose', 'Tercera pose']
    
    fig, axs = plt.subplots(1, 3, figsize=(15, 5))
    
    for i, (pose, title) in enumerate(zip(poses, titles)):
        img = cv2.imread(pose)
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        axs[i].imshow(img_rgb)
        axs[i].set_title(title)
        axs[i].axis('off')  
    
    plt.suptitle('Mis Poses', fontsize=16)
    plt.tight_layout()
    plt.show()

show_poses()
```
