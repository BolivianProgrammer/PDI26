```python
import matplotlib.pyplot as plt
import numpy as np

cell_size = 100
board_size = 8

imagen = np.zeros((board_size * cell_size, board_size * cell_size, 3), dtype=np.uint8)

for i in range(board_size):
    for j in range(board_size):
        y_start = i * cell_size
        y_end = y_start + cell_size
        x_start = j * cell_size
        x_end = x_start + cell_size

        if (i + j) % 2 == 0:
            imagen[y_start:y_end, x_start:x_end, :] = [255, 255, 255] 
        else:
            imagen[y_start:y_end, x_start:x_end, :] = [0, 0, 0]        

plt.figure(figsize=(8, 8))
plt.imshow(imagen)
plt.title('Tablero de ajedrez')
plt.show()
```
