```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

degradado = np.linspace(0, 1, 260).reshape(-1, 1)

plt.imshow(degradado, cmap='viridis', aspect='auto')
plt.title('Degradado Lineal')
plt.show()

degradado_uint8 = (degradado * 255).astype(np.uint8)
degradado_bgr = cv2.applyColorMap(degradado_uint8, cv2.COLORMAP_VIRIDIS)
cv2.imwrite('degradado.jpg', degradado_bgr)
print("Imagen guardada como 'degradado.jpg'")
```
