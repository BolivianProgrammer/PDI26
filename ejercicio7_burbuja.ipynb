# Ejercicio 7: Ordenamiento Burbuja
Ordena números ingresados por teclado a través del método burbuja (Bubble Sort).

```python
def burbuja(lista):
    n = len(lista)
    for i in range(n - 1):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista

try:
    cantidad = int(input("¿Cuántos números desea ordenar? "))
    if cantidad < 2:
        print("Se necesitan al menos 2 números para ordenar.")
    else:
        numeros = []
        for i in range(1, cantidad + 1):
            num = float(input(f"Número {i}: "))
            numeros.append(num)

        print(f"\nLista original:  {numeros}")
        ordenada = burbuja(numeros.copy())
        print(f"Lista ordenada:  {ordenada}")
except ValueError:
    print("Por favor, ingrese valores numéricos válidos.")
```
