Ejercicio 4: Solarizacion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def solarizar(imagen, umbral):
    img_sol = np.where(imagen < umbral, imagen, 255 - imagen)
    return img_sol

degradado = cv2.imread('degradado.jpg')

umbral = 127
degradado_solarizado = solarizar(degradado, umbral)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Degradado Original')
plt.imshow(cv2.cvtColor(degradado, cv2.COLOR_BGR2RGB), aspect='auto')
plt.subplot(1, 2, 2)
plt.title('Degradado Solarizado')
plt.imshow(cv2.cvtColor(degradado_solarizado, cv2.COLOR_BGR2RGB), aspect='auto')
plt.show()
```

Ejercicio 5: Posterizacion

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen, niveles):
    factor_escala = 255 / niveles
    img_pos = np.round(imagen / factor_escala) * factor_escala
    return img_pos.astype(np.uint8)

degradado = cv2.imread('degradado.jpg')

nvl = 3
degradado_posterizado = posterizar(degradado, nvl)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Degradado Original')
plt.imshow(cv2.cvtColor(degradado, cv2.COLOR_BGR2RGB), aspect='auto')
plt.subplot(1, 2, 2)
plt.title('Degradado Posterizado')
plt.imshow(cv2.cvtColor(degradado_posterizado, cv2.COLOR_BGR2RGB), aspect='auto')
plt.show()
```

Falso color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def falso_color(img_gray):
    escal_img = cv2.normalize(img_gray, None, 0, 255, cv2.NORM_MINMAX)
    falso_img = cv2.applyColorMap(escal_img, cv2.COLORMAP_JET)
    return falso_img

degradado = cv2.imread('degradado.jpg')
degradado_gray = cv2.cvtColor(degradado, cv2.COLOR_BGR2GRAY)
degradado_falso = falso_color(degradado_gray)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Degradado Original')
plt.imshow(cv2.cvtColor(degradado, cv2.COLOR_BGR2RGB), aspect='auto')
plt.subplot(1, 2, 2)
plt.title('Degradado con Falso Color')
plt.imshow(cv2.cvtColor(degradado_falso, cv2.COLOR_BGR2RGB), aspect='auto')
plt.show()
```
