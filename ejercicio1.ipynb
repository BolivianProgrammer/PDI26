```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imagenP.jpeg')
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

print(f"Dimensiones de la imagen: {imagen.shape}")

alto, ancho, _ = imagen.shape

x_inicio = int(ancho * 0.28)
x_fin = int(ancho * 0.43)
y_inicio = int(alto * 0.40)
y_fin = int(alto * 0.72)

imagen_recortada = imagen_rgb[y_inicio:y_fin, x_inicio:x_fin]

imagen_recortada_bgr = cv2.cvtColor(imagen_recortada, cv2.COLOR_RGB2BGR)
cv2.imwrite('imagenP_recortada.jpg', imagen_recortada_bgr)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].imshow(imagen_rgb)
axes[0].set_title('Imagen Original')

axes[1].imshow(imagen_recortada)
axes[1].set_title('Imagen Recortada')

plt.tight_layout()
plt.show()
```
