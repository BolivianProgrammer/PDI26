```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imagen1.jpeg')
img_gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

alto, ancho = img_gray.shape
degradado = np.linspace(1, 0, alto).reshape(-1, 1)
degradado_2d = np.tile(degradado, (1, ancho))

img_degradado = (img_gray * degradado_2d).astype(np.uint8)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title('Imagen Original')

axes[1].imshow(img_degradado, cmap='gray')
axes[1].set_title('Imagen con degradado lineal')

plt.tight_layout()
plt.show()
```
