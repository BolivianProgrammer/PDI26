Profundidad de color

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [10,5]

size_image = 100
half = size_image // 2

colors = np.random.randint(0, 256, (4, 3))

imagen = np.zeros((size_image, size_image, 3), dtype=int)

for i in range(4):
    row_start = (i // 2) * half
    col_start = (i % 2) * half
    imagen[row_start:row_start + half, col_start:col_start + half] = colors[i]
    print(f'Color {i}: {colors[i]}')

plt.figure()
plt.imshow(imagen)
plt.title('Colores random')
plt.show()
```

Cuadros con los colores hardcodeados, no random

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [10,5]

size_image = 100
half = size_image // 2

color_top_left     = [255,   0,   0]   
color_top_right    = [  0, 255,   0]   
color_bottom_left  = [  0,   0, 255]   
color_bottom_right = [255, 255,   0]   

imagen_hc = np.zeros((size_image, size_image, 3), dtype=int)
imagen_hc[:half, :half] = color_top_left
imagen_hc[:half, half:] = color_top_right
imagen_hc[half:, :half] = color_bottom_left
imagen_hc[half:, half:] = color_bottom_right

plt.figure()
plt.imshow(imagen_hc)
plt.title('Colores hardcodeados')
plt.show()
```

Imagen 3x3

```python
import matplotlib.pyplot as plt
import numpy as np
size_image = 100
third = size_image // 3

colors = np.random.randint(0, 256, (9, 3))

imagen = np.zeros((size_image, size_image, 3), dtype=int)

for i in range(9):
    row_start = (i // 3) * third
    col_start = (i % 3) * third
    imagen[row_start:row_start + third, col_start:col_start + third] = colors[i]
    print(f'Color {i}: {colors[i]}')

plt.figure()
plt.imshow(imagen)
plt.title('Colores random 3x3')
plt.show()
```

Imagen 3x3 colores hardcodeados

```python
colors_hc = np.array([
    [255, 0, 0],      # Red
    [0, 191, 0],      # Green
    [255, 255, 255],  # White
    [255, 165, 0],    # Orange
    [0, 0, 255],      # Blue
    [255, 255, 0],    # Yellow
])

random_indices = np.random.choice(len(colors_hc), size=9, replace=True)

imagen_hc_3x3 = np.zeros((size_image, size_image, 3), dtype=int)

for i in range(9):
    row_start = (i // 3) * third
    col_start = (i % 3) * third
    imagen_hc_3x3[row_start:row_start + third, col_start:col_start + third] = colors_hc[random_indices[i]]
    print(f'Quadrant {i}: {colors_hc[random_indices[i]]}')

plt.figure()
plt.imshow(imagen_hc_3x3)
plt.title('Colores hardcodeados 3x3 (aleatorio)')
plt.show()
```
