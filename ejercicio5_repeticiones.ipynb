# Ejercicio 5: Número Más Repetido
Permite introducir 7 números por teclado y determina cuál se repite más veces y cuántas.

```python
numeros = []
print("Ingrese 7 números:")
for i in range(1, 8):
    try:
        num = float(input(f"Número {i}: "))
        numeros.append(num)
    except ValueError:
        print("Valor inválido, se asignará 0.")
        numeros.append(0)

conteo = {}
for num in numeros:
    conteo[num] = conteo.get(num, 0) + 1

max_repeticiones = max(conteo.values())
mas_repetidos = [num for num, cnt in conteo.items() if cnt == max_repeticiones]

print(f"\nNúmeros ingresados: {numeros}")
if max_repeticiones == 1:
    print("Ningún número se repite más de una vez.")
elif len(mas_repetidos) == 1:
    print(f"El número que más se repite es: {mas_repetidos[0]} ({max_repeticiones} veces)")
else:
    print(f"Los números que más se repiten son: {mas_repetidos} ({max_repeticiones} veces cada uno)")
```
